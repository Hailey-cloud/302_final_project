```python
cet_france = pd.read_csv('ECB Data Portal_France.csv')
cet_germany = pd.read_csv('ECB Data Portal_Germany.csv')
cet_spain = pd.read_csv('ECB Data Portal_Spain.csv')
```

```python
import pandas as pd


def process_country_df(df, country_name):
    df = df.copy()

 
    original_col = [col for col in df.columns if col.startswith("CET1 capital")][0]
    df = df.rename(columns={original_col: "CET1_capital"})

 
    df['year'] = pd.to_datetime(df['DATE']).dt.year
    df = df[df['year'].isin([2016])]

    
    avg_value = df['CET1_capital'].mean()

    return pd.DataFrame([{
        'Geography': country_name,
        'avg_cet1_capital_2016': avg_value
    }])


agg_spain   = process_country_df(cet_spain, 'Spain')
agg_germany = process_country_df(cet_germany, 'Germany')
agg_france  = process_country_df(cet_france, 'France')


agg_all = pd.concat([agg_spain, agg_germany, agg_france], ignore_index=True)


agg_all.to_csv('cet1_avg_2016.csv', index=False)
```
